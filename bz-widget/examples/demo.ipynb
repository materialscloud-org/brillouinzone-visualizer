# bz-widget
Brillouin zone, k-path and Fermi surfaces in Jupyter. Click a view to rotate it.

```
import seekpath
from bz_widget import BZVisualizer

# Rocksalt NaCl: seekpath's output goes straight in
a = 5.64
structure = (
    [[a, 0, 0], [0, a, 0], [0, 0, a]],
    [[0, 0, 0], [0, .5, .5], [.5, 0, .5], [.5, .5, 0],
     [.5, .5, .5], [.5, 0, 0], [0, .5, 0], [0, 0, .5]],
    [11] * 4 + [17] * 4,
)
BZVisualizer(seekpath.get_explicit_k_path(structure), height="400px")
```

```
# Fermi surfaces from a .bxsf file; options in snake_case
fermi = BZVisualizer(
    bxsf="../../demo/data/simple_example.bxsf",
    energy_range=[-2, 2],
    show_axes=False,
)
fermi
```

```
fermi.fermi_energy, fermi.energy  # E_F from the file; energy follows the slider
```

```
fermi.energy = 9.0  # moves the slider and redraws
```

```
fermi.options = {**fermi.options, "colors": {"zone": "#9ab8d8"}}  # live options
```

```
fermi.view("x")  # or "y", "z", "reset"
```

```
saved = fermi.camera  # updated whenever the view is moved
saved
```

```
fermi.camera = saved  # restore it (also: BZVisualizer(..., camera=saved))
```
